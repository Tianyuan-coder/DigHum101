```python
import pandas as pd
import matplotlib.pyplot as plt
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from collections import Counter
import re
from wordcloud import WordCloud

df = pd.read_csv('UTAustin_2019_sampled.csv')
df['year'] = 2019
df_2022 = pd.read_csv('UTAustin_2022_sampled.csv')
df_2022['year'] = 2022

df_2020 = pd.read_csv('UTAustin_2020_sampled.csv')
df_2020['year'] = 2020

df_2021 = pd.read_csv('UTAustin_2021_sampled.csv')
df_2021['year'] = 2021

combineddf = pd.concat([df, df_2020, df_2021, df_2022], ignore_index=True)
combineddf.to_csv("uta_combined.csv", index=False)
```

```python
# 加载合并后的 Cornell 数据
df = pd.read_csv("uta_combined.csv")  # 修改为你合并后保存的文件名
df = df[df['body'] != '[deleted]']

# 如果尚未有 sentiment_label，可以自动打标签（否则略过这一行）
def assign_sentiment(row):
    sentiments = {
        'positive': row['emo_pred_pos'],
        'negative': row['emo_pred_neg'],
        'neutral': row['emo_pred_neu'],
    }
    return max(sentiments, key=sentiments.get)

if 'sentiment_label' not in df.columns:
    df['sentiment_label'] = df.apply(assign_sentiment, axis=1)


sentiment_pct = df['sentiment_label'].value_counts(normalize=True) * 100
sentiment_pct.plot(kind='pie', autopct='%1.1f%%', startangle=90, 
                   title='Cornell Sentiment Distribution (Percentage)',
                   ylabel='', figsize=(6, 6))
plt.show()


sentiment_by_year = df.groupby(['year', 'sentiment_label']).size().unstack().fillna(0)

# Convert counts to percentages per year
sentiment_by_year_pct = sentiment_by_year.div(sentiment_by_year.sum(axis=1), axis=0) * 100

# Plot
sentiment_by_year_pct.plot(kind='bar', stacked=True, title='Sentiment Over Time (Cornell, %)')
plt.xlabel('Year')
plt.ylabel('Percentage of Comments')
plt.ylim(0, 100)
plt.legend(title='Sentiment')
plt.show()

# 停用词清洗
stop_words = set(stopwords.words('english'))
custom_exclude = {"cornell", "would", "also", "get", "like","people","one","even","much","many","think",'class', 'students',"something","anything","thing","things","really","know","way","make","made","going","go","say","said","see","seen","still","take","taken","want","wanted"}

def get_words(text):
    words = re.findall(r'\b\w+\b', str(text).lower())
    return [word for word in words if word not in stop_words and word not in custom_exclude]

# 提取正负词频
pos_words = Counter()
neg_words = Counter()
for _, row in df.iterrows():
    words = get_words(row['body'])
    if row['sentiment_label'] == 'positive':
        pos_words.update(words)
    elif row['sentiment_label'] == 'negative':
        neg_words.update(words)

# 输出高频词
combined_words = pos_words + neg_words
most_common_combined = set(word for word, _ in combined_words.most_common(10))

for word in most_common_combined:
    pos_count = pos_words[word]
    neg_count = neg_words[word]

pos_top5 = dict(pos_words.most_common(7))
neg_top5 = dict(neg_words.most_common(7))

# 生成词云
def generate_wordcloud(counter, title):
    wc = WordCloud(width=800, height=400).generate_from_frequencies(counter)
    plt.figure(figsize=(10, 5))
    plt.imshow(wc, interpolation='bilinear')
    plt.axis('off')
    plt.title(title)
    plt.show()

generate_wordcloud(pos_top5, "Msu Positive Word Cloud")
generate_wordcloud(neg_top5, "Msu Negative Word Cloud")
```

```python
nrc_df = pd.read_csv("nrc_lexicon.txt", sep="\t", names=["word", "emotion", "association"])
nrc_df = nrc_df[nrc_df["association"] == 1]  

# 构建词 -> 情绪 的映射表
nrc_emotion_dict = defaultdict(list)
for _, row in nrc_df.iterrows():
    nrc_emotion_dict[row["word"]].append(row["emotion"])

# Step 2: 构建 年份 × 情绪 的计数表
emotion_year_matrix = pd.DataFrame()

for _, row in df.iterrows():
    year = row['year']
    words = re.findall(r'\b\w+\b', str(row['body']).lower())
    emotions_this_post = Counter()
    
    for word in words:
        if word in nrc_emotion_dict:
            emotions_this_post.update(nrc_emotion_dict[word])
    
    for emotion, count in emotions_this_post.items():
        #emotion_year_matrix.loc[year, emotion] = emotion_year_matrix.get(emotion, {}).get(year, 0) + count
        current_value = emotion_year_matrix.get(emotion, {}).get(year, 0)
        emotion_year_matrix.loc[year, emotion] = current_value + count


emotion_year_matrix = emotion_year_matrix.fillna(0).astype(int)
emotion_year_pct = emotion_year_matrix.div(emotion_year_matrix.sum(axis=1), axis=0) * 100

def decimal_to_short_label(dyear):
    year = int(dyear) % 100         
    month = int((dyear - int(dyear)) * 12) + 1
    return pd.Timestamp(year=2000+year, month=month, day=1).strftime("%y-%b")


decimal_years = emotion_year_pct.index
short_labels = [decimal_to_short_label(dy) for dy in decimal_years]
emotion_year_pct.index = short_labels

# decimal_years = emotion_year_matrix.index
# short_labels = [decimal_to_short_label(dy) for dy in decimal_years]
# emotion_year_matrix.index = short_labels

ax = emotion_year_pct.plot(kind='line', marker='o', figsize=(14, 6))
plt.title("NRC Emotion Trend by Year (UTA, %)")
plt.xlabel("Year-Month")
plt.ylabel("Percentage of Emotion Words")
plt.legend(title="Emotion", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# 1. Group by year and sentiment_label, then count
sentiment_by_year = df.groupby(['year', 'sentiment_label']).size().unstack().fillna(0)

# 2. Convert raw counts to percentages per year
sentiment_by_year_pct = sentiment_by_year.div(sentiment_by_year.sum(axis=1), axis=0) * 100

# 3. Plot sentiment trends over years
sentiment_by_year_pct.plot(kind='line', marker='o', figsize=(12, 6))
plt.title("Sentiment Percentage Over Time (Cornell)")
plt.xlabel("Year")
plt.ylabel("Percentage of Comments")
plt.ylim(0, 100)
plt.grid(True)
plt.legend(title="Sentiment")
plt.tight_layout()
plt.show()

# 4. Compare Pre-2020 vs Post-2020 Average Sentiment Percentage
pre_covid_avg = sentiment_by_year_pct.loc[lambda df: df.index < 2020].mean()
post_covid_avg = sentiment_by_year_pct.loc[lambda df: df.index >= 2020].mean()

print("📊 Average Sentiment Percentage (Pre-2020):")
print(pre_covid_avg.round(2))

print("\n📊 Average Sentiment Percentage (2020 and After):")
print(post_covid_avg.round(2))
```
